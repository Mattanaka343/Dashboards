```python
import numpy as np
def weekly_data(df, w):

    df = df.sort_values(by='StartTime')
    weeks= df['StartTime'].dt.date.nunique()//6
    excess = df['StartTime'].dt.date.nunique()%6

    dates = df['StartTime'].dt.date.unique()

    f_dates = np.split(dates[:-excess],weeks)
    nef_dates = dates[-excess:]
    print(nef_dates)

    week = np.zeros(len(df))
    
    df['week'] = week

    counter = 1
    for chunk in f_dates:
        for date in chunk:
            for idx in range(len(df)):
                entry = df.loc[idx]
                if entry['StartTime'].date() == date:
                    print('daaa')
                    df.loc[idx,'week'] = counter
        counter += 1

    for date in nef_dates:
        for idx in range(len(df)):
            entry = df.loc[idx]
            if entry['StartTime'].date() == date:
                df.loc[idx,'week'] = counter

    week_db = df #[df['week']==w]

    return week_db
```
